```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

# Đường dẫn đến folder chứa biểu đồ
plot_dir = "runs/detect/doan2_train/"

# Danh sách các biểu đồ bạn cần
plots_needed = {
    "Confusion Matrix":         "confusion_matrix.png",
    "Precision-Recall Curve":   "PR_curve.png",
    "F1-Confidence Curve":      "F1_curve.png",
    "Precision-Confidence Curve": "P_curve.png",
    "Recall-Confidence Curve":    "R_curve.png",
}

# Hiển thị từng biểu đồ
for title, filename in plots_needed.items():
    path = os.path.join(plot_dir, filename)
    if os.path.exists(path):
        img = mpimg.imread(path)
        plt.figure(figsize=(10, 7))
        plt.imshow(img)
        plt.title(title, fontsize=14, fontweight='bold')
        plt.axis('off')
        plt.tight_layout()
        plt.show()
        print(f" hiển thị: {title}")
    else:
        print(f" Không tìm thấy: {filename}")
```
